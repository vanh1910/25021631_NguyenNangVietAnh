```python
# Bài 1: W11A1
with open("sample1.txt", "w") as f:
    f.write("12345")

def reverse(path: str) -> int:
    with open(path, "r") as f:
        data = f.read().strip()
    n = int(data)
    rev = 0
    while n > 0:
        rev = rev * 10 + n % 10
        n //= 10
    return rev

print("Bai 1:", reverse("sample1.txt"))
```

```python
# Bài 20: W11A20
with open("sample20.txt", "w") as f:
    f.write("3\n")
    f.write("1 1 1\n")
    f.write("1 0 1\n")
    f.write("1 1 1\n")

def imageSmoother(path: str):
    with open(path, "r") as f:
        lines = f.read().splitlines()
    m = int(lines[0])
    matrix = []
    for i in range(1, m + 1):
        row = []
        for s in lines[i].split():
            row.append(int(s))
        matrix.append(row)
    result = []
    for i in range(m):
        row = []
        for j in range(m):
            total = 0
            count = 0
            for di in range(-1, 2):
                for dj in range(-1, 2):
                    ni = i + di
                    nj = j + dj
                    if 0 <= ni < m and 0 <= nj < m:
                        total += matrix[ni][nj]
                        count += 1
            row.append(int(total / count))
        result.append(row)
    print("Bai 20:")
    for row in result:
        print(" ".join(str(x) for x in row))

imageSmoother("sample20.txt")
```

```python
# Bài 18: W11A18
with open("sample18.txt", "w") as f:
    f.write("SRC X: 0 Y: 0\n")
    f.write("DES X: 10000 Y: 10000\n")
    f.write("VELOCITY: 6 KM/S\n")

def estimatedTime(path: str) -> float:
    with open(path, "r") as f:
        lines = f.read().splitlines()
    src_parts = lines[0].split()
    src_x = float(src_parts[2])
    src_y = float(src_parts[4])
    des_parts = lines[1].split()
    des_x = float(des_parts[2])
    des_y = float(des_parts[4])
    vel_parts = lines[2].split()
    velocity = float(vel_parts[1])
    dx = des_x - src_x
    dy = des_y - src_y
    distance = (dx * dx + dy * dy) ** 0.5
    time = distance / velocity
    return round(time, 2)

print("Bai 18:", estimatedTime("sample18.txt"))
```

```python
# Bài 17: W11A17
with open("sample17.txt", "w") as f:
    f.write("5 4 6 2 5 2")

def productExceptSelf(path: str) -> list:
    with open(path, "r") as f:
        data = f.read().strip()
    nums = []
    for s in data.split():
        nums.append(int(s))
    n = len(nums)
    result = []
    for i in range(n):
        prod = 1
        for j in range(n):
            if i != j:
                prod *= nums[j]
        result.append(prod)
    return result

print("Bai 17:", productExceptSelf("sample17.txt"))
```

```python
# Bài 16: W11A16
with open("sample16.txt", "w") as f:
    f.write("TYPE: SQUARE\n")
    f.write("SIDE: 5.3\n")
    f.write("\n")
    f.write("TYPE: RECTANGLE\n")
    f.write("WIDTH: 5.0 HEIGHT: 16.0\n")

with open("sample16.txt", "r") as f:
    lines = f.readlines()

i = 0
while i < len(lines):
    line = lines[i].strip()
    if line.startswith("TYPE:"):
        shape_type = line.split(":")[1].strip()
        if shape_type == "SQUARE":
            i += 1
            side_line = lines[i].strip()
            side = float(side_line.split(":")[1].strip())
            perimeter = 4 * side
            print(f"Bai 16: {round(perimeter, 2)}")
        elif shape_type == "RECTANGLE":
            i += 1
            dims_line = lines[i].strip()
            parts = dims_line.split()
            width = float(parts[1])
            height = float(parts[3])
            perimeter = 2 * (width + height)
            print(f"Bai 16: {round(perimeter, 2)}")
    i += 1
```

```python
# Bài 15: W11A15
with open("sample15.csv", "w") as f:
    f.write("id,movie,description,rating\n")
    f.write("1,Inception,A mind bending thriller,9.0\n")
    f.write("2,Movie2,A boring film,8.5\n")
    f.write("3,Interstellar,Space adventure,8.6\n")

def findMovies(path: str) -> list:
    with open(path, "r") as f:
        lines = f.read().splitlines()
    movies = []
    for i in range(1, len(lines)):
        parts = lines[i].split(",")
        if len(parts) >= 4:
            movie = parts[1]
            desc = parts[2]
            rating = float(parts[3])
            if rating > 8.0 and "boring" not in desc.lower():
                movies.append(movie)
    if len(movies) == 0:
        return ["All are boring!"]
    return movies

print("Bai 15:", findMovies("sample15.csv"))
```

```python
# Bài 14: W11A14
with open("sample14.csv", "w") as f:
    f.write("Lap,Time\n")
    f.write("1,1.5\n")
    f.write("2,2.0\n")
    f.write("3,1.8\n")

def averageTime(path: str) -> float:
    with open(path, "r") as f:
        lines = f.read().splitlines()
    total = 0.0
    count = 0
    for i in range(1, len(lines)):
        parts = lines[i].split(",")
        if len(parts) >= 2:
            time = float(parts[1])
            total += time
            count += 1
    avg = total / count
    return round(avg, 2)

print("Bai 14:", averageTime("sample14.csv"))
```

```python
# Bài 13: W11A13
with open("sample13.txt", "w") as f:
    f.write("5\n0 1 0 3 12")

def moveZeroes(fileName: str) -> list:
    with open(fileName, "r") as f:
        lines = f.read().splitlines()
    n = int(lines[0])
    nums = []
    for s in lines[1].split():
        nums.append(int(s))
    result = []
    for num in nums:
        if num != 0:
            result.append(num)
    for num in nums:
        if num == 0:
            result.append(num)
    return result

print("Bai 13:", moveZeroes("sample13.txt"))
```

```python
# Bài 12: W11A12
with open("sample12.txt", "w") as f:
    f.write("4\n1 2 3 4")

def maximumProduct(path: str) -> int:
    with open(path, "r") as f:
        lines = f.read().splitlines()
    n = int(lines[0])
    nums = []
    for s in lines[1].split():
        nums.append(int(s))
    nums_sorted = sorted(nums)
    prod1 = nums_sorted[-1] * nums_sorted[-2] * nums_sorted[-3]
    prod2 = nums_sorted[0] * nums_sorted[1] * nums_sorted[-1]
    if prod1 > prod2:
        return prod1
    else:
        return prod2

print("Bai 12:", maximumProduct("sample12.txt"))
```

```python
# Bài 10: W11A10
with open("sample10.txt", "w") as f:
    f.write("1 2 3 1")

def getMoney(path: str) -> list:
    with open(path, "r") as f:
        data = f.read().strip()
    nums = []
    for s in data.split():
        nums.append(int(s))
    return nums

def rob(houses: list) -> int:
    if len(houses) == 0:
        return 0
    if len(houses) == 1:
        return houses[0]
    prev2 = houses[0]
    prev1 = max(houses[0], houses[1])
    for i in range(2, len(houses)):
        current = max(prev1, prev2 + houses[i])
        prev2 = prev1
        prev1 = current
    return prev1

money = getMoney("sample10.txt")
print("Bai 10:", rob(money))
```

```python
# Bài 6: W11A6
with open("sample6.txt", "w") as f:
    f.write("Hello World")

def countCharacter(path: str, letter: str) -> int:
    with open(path, "r") as f:
        text = f.read()
    cnt = 0
    letter_lower = letter.lower()
    for c in text:
        if c.lower() == letter_lower:
            cnt += 1
    return cnt

print("Bai 6:", countCharacter("sample6.txt", "l"))
```

```python
# Bài 5: W11A5
with open("sample5.txt", "w") as f:
    f.write("We accept the love we think we deserve.")

def count(path: str, character: str) -> int:
    with open(path, "r") as f:
        text = f.read()
    cnt = 0
    for c in text:
        if c == character:
            cnt += 1
    return cnt

print("Bai 5:", count("sample5.txt", "e"))
```

```python
# Bài 4: W11A4
with open("sample4a.txt", "w") as f:
    f.write("Tom Marvolo Riddle")
with open("sample4b.txt", "w") as f:
    f.write("I Am Lord Voldemort")

def areAnagrams(path1: str, path2: str) -> bool:
    with open(path1, "r") as f:
        s1 = f.read()
    with open(path2, "r") as f:
        s2 = f.read()
    s1_clean = ""
    for c in s1.lower():
        if c.isalpha():
            s1_clean += c
    s2_clean = ""
    for c in s2.lower():
        if c.isalpha():
            s2_clean += c
    if len(s1_clean) != len(s2_clean):
        return False
    count1 = {}
    for c in s1_clean:
        count1[c] = count1.get(c, 0) + 1
    count2 = {}
    for c in s2_clean:
        count2[c] = count2.get(c, 0) + 1
    return count1 == count2

print("Bai 4:", areAnagrams("sample4a.txt", "sample4b.txt"))
```

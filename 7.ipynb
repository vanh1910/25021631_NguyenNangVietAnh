```python
#W7A1
def binary_search(arr, left, right, target):
    while left <= right:
        mid = (left + right) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    return -1
```

```python
# Test W7A1
print("W7A1 - Binary Search Tests:")
print(binary_search([1, 2, 7], 0, 2, -1))  # Expected: -1
print(binary_search([1, 2, 3, 4, 5, 6, 7, 8], 0, 7, 3))  # Expected: 2
print(binary_search([1, 3, 4, 6, 7, 9, 100, 1000], 0, 7, 1000))  # Expected: 7
print(binary_search([1, 3, 4, 60, 70, 90, 100, 1000], 0, 7, 60))  # Expected: 3
```

```python
#W7A3
def selection_sort(arr):
    n = len(arr)
    for i in range(n):
        min_idx = i
        for j in range(i+1, n):
            if arr[j] < arr[min_idx]:
                min_idx = j
        arr[i], arr[min_idx] = arr[min_idx], arr[i]
    return arr
```

```python
# Test W7A3
print("W7A3 - Selection Sort Tests:")
print(selection_sort([6, 7, 8, 1, 5, 2, 3, 4]))  # Expected: [1, 2, 3, 4, 5, 6, 7, 8]
print(selection_sort([1, 3, 4, 6, 7, 9, 100, 1000]))  # Expected: [1, 3, 4, 6, 7, 9, 100, 1000]
print(selection_sort([70, 90, 100, 1, 3, 4, 60, 1000]))  # Expected: [1, 3, 4, 60, 70, 90, 100, 1000]
```

```python
#W7A4
def most_frequent(arr):
    counts = {}
    first_idx = {}
    for i, v in enumerate(arr):
        counts[v] = counts.get(v, 0) + 1
        if v not in first_idx:
            first_idx[v] = i
    best = None
    best_count = 0
    best_first = float('inf')
    for v, c in counts.items():
        if c > best_count or (c == best_count and first_idx[v] < best_first):
            best = v
            best_count = c
            best_first = first_idx[v]
    return best, best_count
```

```python
# Test W7A4
print("W7A4 - Most Frequent Tests:")
val, count = most_frequent([2, 3, 4, 2, 2, 5])
print(f"{val} xuat hien nhieu nhat, som nhat, {count} lan")  # Expected: 2, 3 lan
val, count = most_frequent([6, 8, 1, 2, 3, 10, 6, 4, 4, 12, 13])
print(f"{val} xuat hien nhieu nhat, som nhat, {count} lan")  # Expected: 6, 2 lan
val, count = most_frequent([1, 2, 2, 1, 2, 1, 10, 12, 12, 1, 10, 10, 10])
print(f"{val} xuat hien nhieu nhat, som nhat, {count} lan")  # Expected: 1, 4 lan
val, count = most_frequent([1, 1, 1, 1, 2, 2, 2, 2])
print(f"{val} xuat hien nhieu nhat, som nhat, {count} lan")  # Expected: 1, 4 lan
```

```python
#W7A5
def count_pairs_with_sum(arr, X):
    from collections import Counter
    cnt = Counter(arr)
    total = 0
    for v in list(cnt.keys()):
        complement = X - v
        if complement in cnt:
            if complement == v:
                total += cnt[v] * (cnt[v] - 1) // 2
            elif complement > v:
                total += cnt[v] * cnt[complement]
    return total
```

```python
# Test W7A5
print("W7A5 - Count Pairs with Sum Tests:")
print(count_pairs_with_sum([1, 5, 7, -1, 5], 6))  # Expected: 2
```

```python
#W7A6
def flatten(nested):
    return [x for sub in nested for x in sub]
```

```python
#W7A7
def length_of_LIS(nums):
    import bisect
    tails = []
    for x in nums:
        i = bisect.bisect_left(tails, x)
        if i == len(tails):
            tails.append(x)
        else:
            tails[i] = x
    return len(tails)
```

```python
# Test W7A7
print("W7A7 - Length of LIS Tests:")
print(length_of_LIS([10, 9, 2, 5, 3, 7, 101, 18]))  # Expected: 4
print(length_of_LIS([0, 1, 0, 3, 2, 3]))  # Expected: 4
print(length_of_LIS([7, 7, 7, 7, 7, 7, 7]))  # Expected: 1
```

```python
#W7A8
def min_interval_cover(intervals, queries):
    res = []
    for q in queries:
        best = None
        for l, r in intervals:
            if l <= q <= r:
                length = r - l + 1
                if best is None or length < best:
                    best = length
        res.append(best if best is not None else -1)
    return res
```

```python
# Test W7A8
print("W7A8 - Min Interval Cover Tests:")
print(min_interval_cover([[1, 4], [2, 4], [3, 6], [4, 4]], [2, 3, 4, 5]))  # Expected: [3, 3, 1, 3]
print(min_interval_cover([[2, 8], [1, 5], [10, 15]], [1, 4, 6, 11, 20]))  # Expected: [5, 4, 7, 6, -1]
```

```python
#W7A9
def count_smaller(nums):
    import bisect
    sorted_list = []
    res = []
    for x in reversed(nums):
        i = bisect.bisect_left(sorted_list, x)
        res.append(i)
        bisect.insort_left(sorted_list, x)
    return list(reversed(res))
```

```python
# Test W7A9
print("W7A9 - Count Smaller Tests:")
print(count_smaller([5, 2, 6, 1]))  # Expected: [2, 1, 1, 0]
print(count_smaller([3, 2, 2, 6, 1]))  # Expected: [3, 1, 1, 1, 0]
print(count_smaller([1, 2, 3, 4, 5]))  # Expected: [0, 0, 0, 0, 0]
print(count_smaller([5, 4, 3, 2, 1]))  # Expected: [4, 3, 2, 1, 0]
```

```python
#W7A10
def max_subarray_sum_modulo(nums, m):
    import bisect
    prefix = 0
    best = 0
    prefixes = []
    for num in nums:
        prefix = (prefix + num) % m
        best = max(best, prefix)
        i = bisect.bisect_right(prefixes, prefix)
        if i < len(prefixes):
            candidate = (prefix - prefixes[i] + m) % m
            best = max(best, candidate)
        bisect.insort(prefixes, prefix)
    return best
```

```python
# Test W7A10
print("W7A10 - Max Subarray Sum Modulo Tests:")
print(max_subarray_sum_modulo([3, 3, 9, 9, 5], 7))  # Expected: 6
print(max_subarray_sum_modulo([1, 2, 3], 2))  # Expected: 1
print(max_subarray_sum_modulo([10, 20, 30], 15))  # Expected: 10
```

```python
#W7A11
def count_word_occurrences(text, pattern):
    import re
    return len(re.findall(r'\b' + re.escape(pattern) + r'\b', text, flags=re.IGNORECASE))
```

```python
# Test W7A11
print("W7A11 - Count Word Occurrences Tests:")
print(count_word_occurrences("There's a fire that burns inside.", "fire"))  # Expected: 1
print(count_word_occurrences("It's an instinct that never lies. The target's tattooed between our eyes.", "never lies"))  # Expected: 1
print(count_word_occurrences("There's only two options. Win or die, win or die. Win or die, die, die, die, die.", "die"))  # Expected: 7
```

```python
#W7A12
def sort_labels(phrases):
    labels = ['A','B','C','D']
    paired = list(zip(phrases, labels))
    paired.sort(key=lambda x: x[0].lower())
    return [label for _, label in paired]
```

```python
# Test W7A12
print("W7A12 - Sort Labels Tests:")
print(' '.join(sort_labels(['May', 'Troi', 'Cua', 'Em'])))  # Expected: C D A B
print(' '.join(sort_labels(['Hoa hong', 'hoa hue', 'hoa lan', 'hoa huong duong'])))  # Expected: A B D C
```

```python
#W7A13
def longest_harmonious_subsequence(arr):
    from collections import Counter
    cnt = Counter(arr)
    best = 0
    for k in cnt:
        if k + 1 in cnt:
            best = max(best, cnt[k] + cnt[k+1])
    return best
```

```python
# Test W7A13
print("W7A13 - Longest Harmonious Subsequence Tests:")
print(longest_harmonious_subsequence([1, 2, 3, 4]))  # Expected: 2
print(longest_harmonious_subsequence([1, 3, 2, 2, 5, 2, 3, 7]))  # Expected: 5
```

```python
#W7A14
def find_sevens(arr):
    idxs = [i for i,v in enumerate(arr) if v == 7]
    if not idxs:
        return 'Not found'
    return sorted(idxs, reverse=True)
```

```python
# Test W7A14
print("W7A14 - Find Sevens Tests:")
print(find_sevens([1, 7, 10, -2, 12]))  # Expected: 1
print(find_sevens([-12, -5, -4, -12, -14, 5, 10, -8, 4, -3]))  # Expected: Not found
```

```python
#W7A15
def nemo_neighbors(names):
    n = len(names)
    idx = names.index('Nemo')
    return f"{names[(idx-1)%n]} and {names[(idx+1)%n]}"
```

```python
# Test W7A15
print("W7A15 - Nemo Neighbors Tests:")
print(nemo_neighbors(['Anastatia', 'Alice', 'Ariel', 'Aurora', 'Nemo']))  # Expected: Aurora and Anastatia
```
